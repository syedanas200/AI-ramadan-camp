```python
Eng = int(input("enter your Eng makrs:"))
Sci = int(input("enter your Sci makrs:"))
Maths = int(input("enter your Maths makrs:"))
Urdu = int(input("enter your Urdu makrs:"))
Isl = int(input("enter your Isl makrs:"))
obtain = Eng+Sci+Maths+Urdu+Isl
print("your obtain marks is:",obtain)
perc = obtain/500*100

if perc >=80 and perc <=100:
    print("A-one grade")
elif perc >=70 and perc <=80:
    print("A grade")
elif perc >=60 and perc <=70:
    print("B grade")
elif perc >=50 and perc <=60:
    print("C grade")
elif perc >=40 and perc <=50:
    print("D grade")
elif perc >=33 and perc <=40:
    print("E grade")
else:
    print("fail")
```
