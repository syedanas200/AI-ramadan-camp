```python
input1 = int(input("enter value1 :"))
input2 = int(input("enter value2 :"))
operand = input("enter operand :")
if operand == "+":
    print(input1 + input2)
elif operand == "-":
    print(input1 - input2)
elif operand == "*":
    print(input1 * input2)
elif operand == "/":
    print(input1 / input2)
elif operand == "**":
    print(input1 ** input2)
else:
    print("please enter correct operand")
```

```python
my_list = ["apple", "banana", "25", "orange", "99"]

has_number = False

for item in my_list:
    if isinstance(item, (int, float)):
        has_number = True
        break
    if isinstance(item, str):
        try:
            float(item)  
            has_number = True
            break
        except ValueError:
            pass

if has_number:
    print("The list contains a numeric value.")
else:
    print("The list does not contain any numeric value.")
```
